```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, classification_report

# Download NLTK data
nltk.download('vader_lexicon')

# Sample data (replace with your own dataset)
data = [
    {'text': 'The product is excellent and works perfectly.', 'label': 'positive'},
    {'text': 'The device is faulty and doesn\'t function well.', 'label': 'negative'},
    # Add more examples with different labels
]

# Preprocess the data
texts = [item['text'] for item in data]
labels = [item['label'] for item in data]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=42)

# Vectorize the text data
vectorizer = CountVectorizer()
X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

# Train a simple random forest classifier
clf = RandomForestClassifier(random_state=42)
clf.fit(X_train_vectorized, y_train)

# Make predictions on the test set
predictions = clf.predict(X_test_vectorized)

# Evaluate the model
accuracy = accuracy_score(y_test, predictions)
report = classification_report(y_test, predictions)

print(f"Accuracy: {accuracy:.2f}")
print("Classification Report:\n", report)

# Sentiment analysis using NLTK
sia = SentimentIntensityAnalyzer()

# Example of sentiment analysis for a new text
new_text = "The article provides valuable insights and is well-written."
sentiment_score = sia.polarity_scores(new_text)['compound']

print(f"Sentiment Score: {sentiment_score:.2f}")
```
